```python
import matplotlib.pyplot as plt
%matplotlib inline

import seaborn as sns
sns.set(color_codes=True)

import datetime
```

```python
from scipy.stats import bernoulli
```

```python
data_bern = bernoulli.rvs(size=10000, p=0.6)

ax = sns.displot(data_bern,
                  bins=100,
                  kde=False,
                 )
ax.set(xlabel='Bernoulli Distribution ', ylabel='Frequency')
plt.show()
```

```python
from scipy.stats import binom
```

```python
data_binom = binom.rvs(n=10, p=0.6, size=1000)
ax = sns.displot(data_binom,
                  bins=100,
                  kde=False,
                 )
ax.set(xlabel='Binomial Distribution ', ylabel='Frequency')
plt.show()
```

```python
from scipy.stats import poisson
```

```python
data_poisson = poisson.rvs(mu=3, size=10000)
ax = sns.histplot(data_poisson,
                  bins=50,
                  kde=False,
                 )
ax.set(xlabel='Poisson Distribution ', ylabel='Frequency')
plt.show()
```

```python
import numpy as np
np.random.seed(1337)

num_bins = 50  # number of bins in each dimension  Try: 5, 20, 50, 500 
num_samples = 10000000

data = 100 + 10 * np.random.randn(num_samples, 2)  # generate 100000 2-d random data points
joint_probs, edges = np.histogramdd(data, bins=num_bins)

edges = np.array(edges)[:,1:]

import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

%matplotlib inline
plt.style.use('ggplot')

X = edges[0]
Y = edges[1]
X, Y = np.meshgrid(X, Y)

Z = joint_probs / joint_probs.sum()

fig = plt.figure()
ax = plt.axes(projection='3d')

ax.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap='viridis')

ax.view_init(45, 30)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```
