### Quantifying Chances
    - Flip a coin:
        Q. What are the chances that it will land on its head? (Hard)
        Q. If I flip it 10 times, how many will be heads? (Easy)
        
        Ω = {H, T} -> events 
        both are equally likely, 50% chance of each
        P(H)=0.5 , P(T) = 0.5

    - The two axioms of Probability:
        must lie b/w [0, 1] 
        sum of probability of all events must be 1
      
    - Flip the coin twice:
        Ω = {HH, HT, TH, TT} , both flips are "independent"
        P(H) = 25%

    - P(Event) = Favourable Outcomes / Total outcomes

```python
import matplotlib 
import matplotlib.pyplot as plt
%matplotlib inline

import numpy as np

import seaborn as sns
sns.set(color_codes=True)
sns.set_style("white")

np.random.seed(0)

# generate a 'flip'
def flip(num = 1):
    flips = []

    for i in range(num):
        num = np.random.uniform(low=0.0, high=1.0)
        if num > 0.5:
            flips.append('H')   # should be doing yield here if you know 'generators'
        else:
            flips.append('T')
    return flips

flips = flip(10)
values, counts = np.unique(flips, return_counts=True)

print(flips)
print(counts)
```

### Probability of Flips 

```python
from collections import Counter, defaultdict

def get_freqs(flips):
    key = Counter(flips).keys()
    vals = Counter(flips).values()

    # return dict(zip(key, vals))    # bus: what if there are no 'H' or no 'T' (all are Head or all are Tails)
    
    return defaultdict(int, dict(zip(key, vals)))
    
```

```python
freqs = get_freqs(flips)
print(freqs)
```

```python
prob_h = freqs['H'] / len(flips)
print(prob_h)
```

### Experiment: Probability calculated based on 1 flip unto N flips

```python
maximum_flips = 1000

probs = []
for num_flips in range(1, maximum_flips):
    flips = flip(num_flips)
    freqs = get_freqs(flips)
    prob_h = freqs['H'] / len(flips)

    probs.append(prob_h)

# print(probs)
```

```python
print(freqs)
```

```python
plt.plot(probs)
plt.ylabel('Estimated Probability')
plt.xlabel('Number of Flips in Experiment')
plt.title('Calculating Proabability over Number of Flips')
sns.despine(offset=10, trim=True) # move axes away
plt.show()
```

```python
plt.plot(probs[:10])
plt.ylabel('Estimated Probability')
plt.xlabel('Number of Flips in Experiment')
plt.title('Calculating Proabability over Number of Flips')
sns.despine(offset=10, trim=True) # move axes away
plt.show()
```

```python
plt.plot(probs[maximum_flips-10:])    # last 10 experiments
plt.ylabel("Estimated Probability")
plt.xlabel('Number of Flips in Experiment')
plt.title('Calculating Proabability over Number of Flips')
sns.despine(offset=10, trim=True) # move axes away
plt.show()
```
