```python
test_sentence = non[4]
test_sentence = non[5]
test_sentence = spam[1]

print(test_sentence)

removed_stops = remove_stopwards(test_sentence)
print(test_stops)

p = PorterStemmer()
stemmed = p.stem(removed_stops)
print(stemmed)

tokens = tokensize(stemmed)
print(list(tokens))

```
