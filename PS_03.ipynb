# Statistics
    Academic discipline dealing with all aspects of data
    
### Perspectives:
    - art of summerizing data -> make data comprehensible
    - science of uncertainty -> most info in the world in uncertain
    - science of decisions -> ultimate goal of statistics
    - science of variation -> central tendency and spread
    - art of forecasting -> predictions
    - science of measurement and data collection

### Foundations of data
#### Sources of data
    - Designed data -> "artificially collected" (surveys, studies etc)
    - Organic data (process generated)
    - for both, data needs to be IID -> independant, identically distributed
#### Types of data
    - Weight -> numeric, continuos
    - No. of kids -> numeric, discrete
    - Age group (child, adult, elder) -> categorical, ordered
    - Gender -> categorical, unordered
    
    Practical Note:
        Gender represented as: M / F
                           or: 0 / 1 -> still unordered but thought of as ordered
        M [ 1 0 ] 
        F [ 0 1 ]
        "one-hot vector representation" 
        
### Variables
    |__Quantitative(numeric)
    |  |__Continuous
    |  |__Discrete
    |   
    |__Qualitative(categorical)
       |__Ordinal(ranked)
       |__Nominal(unranked)
